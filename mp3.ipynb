```python
from pathlib import Path
import yt_dlp

url = "https://www.youtube.com/watch?v=Dpo-kYbeUQ4&list=RDDpo-kYbeUQ4&start_radio=1"

# กำหนดโฟลเดอร์สำหรับเก็บไฟล์
download_folder = Path(r"D:\YouTube-MP3")
download_folder.mkdir(parents=True, exist_ok=True)

options = {
    "format": "bestaudio/best",
    "outtmpl": str(download_folder / "%(title)s.%(ext)s"),

    "postprocessors": [
        {
            "key": "FFmpegExtractAudio",
            "preferredcodec": "mp3",
            "preferredquality": "320",
        }
    ],
}

with yt_dlp.YoutubeDL(options) as ydl:
    ydl.download([url])

print("\nดาวน์โหลดเสร็จแล้ว!")
print(f"ไฟล์อยู่ที่: {download_folder}")
```
