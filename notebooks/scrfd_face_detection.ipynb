```python
# ============================================================
# scrfd_face_detection.py
#
# Standalone Face Detection Pipeline
# Lip Reading / Visual Speech Recognition
#
# Model:
#   InsightFace + SCRFD (buffalo_l)
#
# Input:
#   dataset/raw_videos
#
# Output:
#   dataset/scrfd_detection
#
# ============================================================

from pathlib import Path
import argparse
import csv
import json
import sys
import math

import cv2
import numpy as np


# ============================================================
# CHECK DEPENDENCIES
# ============================================================

try:
    import onnxruntime as ort
except ImportError:
    print(
        "\n[ERROR] ไม่พบ onnxruntime\n"
        "ติดตั้งด้วย:\n"
        "python -m pip install onnxruntime-gpu\n"
    )
    sys.exit(1)

try:
    from insightface.app import FaceAnalysis
except ImportError:
    print(
        "\n[ERROR] ไม่พบ insightface\n"
        "ติดตั้งด้วย:\n"
        "python -m pip install insightface\n"
    )
    sys.exit(1)


# ============================================================
# PROJECT PATH
# ============================================================

SCRIPT_DIR = Path(__file__).resolve().parent
PROJECT_ROOT = SCRIPT_DIR.parent

RAW_VIDEO_DIR = (
    PROJECT_ROOT
    / "dataset"
    / "raw_videos"
)

OUTPUT_ROOT = (
    PROJECT_ROOT
    / "dataset"
    / "scrfd_detection"
)


SUPPORTED_VIDEO_EXTENSIONS = {
    ".mp4",
    ".webm",
    ".avi",
    ".mov",
    ".mkv",
    ".mpeg",
    ".mpg",
}


# ============================================================
# UTILITY
# ============================================================

def log(message=""):
    print(message, flush=True)


def discover_videos(video_dir: Path):

    if not video_dir.exists():
        raise FileNotFoundError(
            f"ไม่พบโฟลเดอร์:\n{video_dir}"
        )

    videos = [
        p
        for p in video_dir.iterdir()
        if p.is_file()
        and p.suffix.lower()
        in SUPPORTED_VIDEO_EXTENSIONS
    ]

    return sorted(videos)


# ============================================================
# IoU
# ============================================================

def calculate_iou(box_a, box_b):

    ax1, ay1, ax2, ay2 = box_a
    bx1, by1, bx2, by2 = box_b

    x1 = max(ax1, bx1)
    y1 = max(ay1, by1)

    x2 = min(ax2, bx2)
    y2 = min(ay2, by2)

    inter_w = max(0, x2 - x1)
    inter_h = max(0, y2 - y1)

    intersection = inter_w * inter_h

    area_a = max(0, ax2 - ax1) * max(
        0,
        ay2 - ay1
    )

    area_b = max(0, bx2 - bx1) * max(
        0,
        by2 - by1
    )

    union = (
        area_a
        + area_b
        - intersection
    )

    if union <= 0:
        return 0.0

    return intersection / union


# ============================================================
# CENTER DISTANCE
# ============================================================

def center_distance(box_a, box_b):

    ax = (box_a[0] + box_a[2]) / 2
    ay = (box_a[1] + box_a[3]) / 2

    bx = (box_b[0] + box_b[2]) / 2
    by = (box_b[1] + box_b[3]) / 2

    return math.sqrt(
        (ax - bx) ** 2
        +
        (ay - by) ** 2
    )


# ============================================================
# SIMPLE FACE TRACKER
#
# ใช้ IoU + center distance
# เพื่อรักษา track ID ระหว่างเฟรม
#
# ยังไม่ใช่ DeepSORT / ByteTrack
# แต่เพียงพอสำหรับทดสอบ preprocessing
# ============================================================

class SimpleFaceTracker:

    def __init__(
        self,
        iou_threshold=0.25,
        max_distance=150,
        max_missing=15,
    ):

        self.iou_threshold = (
            iou_threshold
        )

        self.max_distance = (
            max_distance
        )

        self.max_missing = (
            max_missing
        )

        self.next_track_id = 0

        self.tracks = {}


    def update(self, detections):

        # ----------------------------------------
        # Mark all tracks as missing
        # ----------------------------------------

        for track_id in self.tracks:

            self.tracks[
                track_id
            ]["missing"] += 1

        assigned = []

        # ----------------------------------------
        # Match detection -> existing track
        # ----------------------------------------

        for detection in detections:

            bbox = detection["bbox"]

            best_track_id = None
            best_score = -1

            for (
                track_id,
                track
            ) in self.tracks.items():

                previous_bbox = (
                    track["bbox"]
                )

                iou = calculate_iou(
                    bbox,
                    previous_bbox
                )

                distance = (
                    center_distance(
                        bbox,
                        previous_bbox
                    )
                )

                # --------------------------------
                # Matching score
                # --------------------------------

                if (
                    iou
                    >= self.iou_threshold
                    or
                    distance
                    <= self.max_distance
                ):

                    score = (
                        iou
                        -
                        distance
                        / 10000.0
                    )

                    if score > best_score:

                        best_score = score
                        best_track_id = (
                            track_id
                        )

            # ----------------------------------------
            # Existing track
            # ----------------------------------------

            if best_track_id is not None:

                detection[
                    "track_id"
                ] = best_track_id

                self.tracks[
                    best_track_id
                ] = {
                    "bbox": bbox,
                    "missing": 0,
                }

            # ----------------------------------------
            # New track
            # ----------------------------------------

            else:

                track_id = (
                    self.next_track_id
                )

                self.next_track_id += 1

                detection[
                    "track_id"
                ] = track_id

                self.tracks[
                    track_id
                ] = {
                    "bbox": bbox,
                    "missing": 0,
                }

            assigned.append(
                detection
            )

        # ----------------------------------------
        # Remove stale tracks
        # ----------------------------------------

        stale_tracks = [

            track_id

            for (
                track_id,
                track
            ) in self.tracks.items()

            if (
                track["missing"]
                > self.max_missing
            )
        ]

        for track_id in stale_tracks:

            del self.tracks[
                track_id
            ]

        return assigned


# ============================================================
# INITIALIZE SCRFD
# ============================================================

def create_face_detector(
    det_size=640,
    force_cpu=False
):

    available_providers = (
        ort.get_available_providers()
    )

    log("")
    log("=" * 70)
    log("ONNX Runtime")
    log("=" * 70)

    log(
        f"Available providers: "
        f"{available_providers}"
    )

    # --------------------------------------------
    # GPU
    # --------------------------------------------

    if (
        not force_cpu
        and
        "CUDAExecutionProvider"
        in available_providers
    ):

        providers = [
            "CUDAExecutionProvider",
            "CPUExecutionProvider",
        ]

        ctx_id = 0

        log(
            "Using CUDAExecutionProvider"
        )

    # --------------------------------------------
    # CPU fallback
    # --------------------------------------------

    else:

        providers = [
            "CPUExecutionProvider"
        ]

        ctx_id = -1

        log(
            "Using CPUExecutionProvider"
        )

    log("=" * 70)

    # ============================================
    # buffalo_l includes SCRFD detector
    # ============================================

    app = FaceAnalysis(
        name="buffalo_l",
        providers=providers,
    )

    app.prepare(
        ctx_id=ctx_id,
        det_size=(
            det_size,
            det_size
        ),
    )

    return app


# ============================================================
# SAVE VIDEO METADATA
# ============================================================

def save_video_metadata(
    output_dir,
    metadata
):

    path = (
        output_dir
        / "video_metadata.json"
    )

    with open(
        path,
        "w",
        encoding="utf-8"
    ) as f:

        json.dump(
            metadata,
            f,
            ensure_ascii=False,
            indent=4,
        )


# ============================================================
# PROCESS VIDEO
# ============================================================

def process_video(
    video_path,
    detector,
    output_root,
    max_frames=None,
    frame_stride=1,
    preview_every=100,
    confidence_threshold=0.5,
):

    video_id = (
        video_path.stem
    )

    video_output_dir = (
        output_root
        / video_id
    )

    preview_dir = (
        video_output_dir
        / "preview"
    )

    video_output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    preview_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    csv_path = (
        video_output_dir
        / "detection.csv"
    )

    # ============================================
    # Open video
    # ============================================

    cap = cv2.VideoCapture(
        str(video_path)
    )

    if not cap.isOpened():

        raise RuntimeError(
            f"เปิดวิดีโอไม่ได้: "
            f"{video_path}"
        )

    fps = float(
        cap.get(
            cv2.CAP_PROP_FPS
        )
    )

    total_frames = int(
        cap.get(
            cv2.CAP_PROP_FRAME_COUNT
        )
    )

    width = int(
        cap.get(
            cv2.CAP_PROP_FRAME_WIDTH
        )
    )

    height = int(
        cap.get(
            cv2.CAP_PROP_FRAME_HEIGHT
        )
    )

    duration = (
        total_frames / fps
        if fps > 0
        else 0
    )

    log("")
    log("=" * 70)

    log(
        f"VIDEO: {video_id}"
    )

    log("=" * 70)

    log(
        f"Resolution : "
        f"{width}x{height}"
    )

    log(
        f"FPS        : "
        f"{fps:.3f}"
    )

    log(
        f"Frames     : "
        f"{total_frames}"
    )

    log(
        f"Duration   : "
        f"{duration:.2f}s"
    )

    log("=" * 70)

    metadata = {

        "video_id":
            video_id,

        "filename":
            video_path.name,

        "fps":
            fps,

        "width":
            width,

        "height":
            height,

        "total_frames":
            total_frames,

        "duration_seconds":
            duration,

    }

    save_video_metadata(
        video_output_dir,
        metadata
    )

    # ============================================
    # CSV
    # ============================================

    csv_file = open(
        csv_path,
        "w",
        newline="",
        encoding="utf-8-sig"
    )

    writer = csv.writer(
        csv_file
    )

    writer.writerow([

        "video_id",

        "frame_index",

        "timestamp_seconds",

        "track_id",

        "face_index",

        "x1",
        "y1",
        "x2",
        "y2",

        "width",

        "height",

        "det_score",

        "left_eye_x",
        "left_eye_y",

        "right_eye_x",
        "right_eye_y",

        "nose_x",
        "nose_y",

        "left_mouth_x",
        "left_mouth_y",

        "right_mouth_x",
        "right_mouth_y",

    ])

    # ============================================
    # Tracker
    # ============================================

    tracker = (
        SimpleFaceTracker()
    )

    source_frame_index = 0
    processed_frames = 0
    total_faces = 0

    # ============================================
    # LOOP
    # ============================================

    try:

        while True:

            ret, frame = cap.read()

            if not ret:
                break

            # ------------------------------------
            # Frame stride
            # ------------------------------------

            if (
                source_frame_index
                % frame_stride
                != 0
            ):

                source_frame_index += 1
                continue

            # ------------------------------------
            # SCRFD detection
            # ------------------------------------

            faces = detector.get(
                frame
            )

            detections = []

            for face in faces:

                score = float(
                    face.det_score
                )

                if (
                    score
                    < confidence_threshold
                ):
                    continue

                bbox = (
                    face.bbox
                    .astype(float)
                    .tolist()
                )

                kps = None

                if (
                    hasattr(
                        face,
                        "kps"
                    )
                    and
                    face.kps
                    is not None
                ):

                    kps = (
                        face.kps
                        .astype(float)
                        .tolist()
                    )

                detections.append({

                    "bbox":
                        bbox,

                    "score":
                        score,

                    "kps":
                        kps,

                })

            # ------------------------------------
            # Tracking
            # ------------------------------------

            detections = (
                tracker.update(
                    detections
                )
            )

            # ------------------------------------
            # Timestamp
            # ------------------------------------

            timestamp = (

                source_frame_index
                / fps

                if fps > 0

                else 0
            )

            # ------------------------------------
            # Draw preview
            # ------------------------------------

            preview_frame = (
                frame.copy()
            )

            # ------------------------------------
            # Save detections
            # ------------------------------------

            for face_index, det in enumerate(
                detections
            ):

                (
                    x1,
                    y1,
                    x2,
                    y2
                ) = det["bbox"]

                face_width = (
                    x2 - x1
                )

                face_height = (
                    y2 - y1
                )

                score = (
                    det["score"]
                )

                track_id = (
                    det["track_id"]
                )

                kps = (
                    det["kps"]
                )

                # --------------------------------
                # 5 landmarks
                # --------------------------------

                landmarks = [
                    "",
                    "",
                    "",
                    "",
                    "",
                    "",
                    "",
                    "",
                    "",
                    "",
                ]

                if (
                    kps is not None
                    and
                    len(kps) >= 5
                ):

                    landmarks = [

                        kps[0][0],
                        kps[0][1],

                        kps[1][0],
                        kps[1][1],

                        kps[2][0],
                        kps[2][1],

                        kps[3][0],
                        kps[3][1],

                        kps[4][0],
                        kps[4][1],

                    ]

                writer.writerow([

                    video_id,

                    source_frame_index,

                    round(
                        timestamp,
                        6
                    ),

                    track_id,

                    face_index,

                    round(x1, 3),
                    round(y1, 3),
                    round(x2, 3),
                    round(y2, 3),

                    round(
                        face_width,
                        3
                    ),

                    round(
                        face_height,
                        3
                    ),

                    round(
                        score,
                        6
                    ),

                    *landmarks,

                ])

                total_faces += 1

                # --------------------------------
                # Bounding box
                # --------------------------------

                cv2.rectangle(

                    preview_frame,

                    (
                        int(x1),
                        int(y1)
                    ),

                    (
                        int(x2),
                        int(y2)
                    ),

                    (
                        0,
                        255,
                        0
                    ),

                    2

                )

                # --------------------------------
                # Track label
                # --------------------------------

                label = (
                    f"ID {track_id} "
                    f"{score:.2f}"
                )

                cv2.putText(

                    preview_frame,

                    label,

                    (
                        int(x1),
                        max(
                            20,
                            int(y1) - 10
                        )
                    ),

                    cv2.FONT_HERSHEY_SIMPLEX,

                    0.7,

                    (
                        0,
                        255,
                        0
                    ),

                    2,

                    cv2.LINE_AA,

                )

                # --------------------------------
                # Landmark preview
                # --------------------------------

                if kps is not None:

                    for point in kps:

                        px = int(
                            point[0]
                        )

                        py = int(
                            point[1]
                        )

                        cv2.circle(

                            preview_frame,

                            (
                                px,
                                py
                            ),

                            3,

                            (
                                0,
                                0,
                                255
                            ),

                            -1

                        )

            # ====================================
            # SAVE PREVIEW
            # ====================================

            if (
                preview_every > 0
                and
                processed_frames
                % preview_every
                == 0
            ):

                preview_path = (

                    preview_dir

                    / (
                        f"frame_"
                        f"{source_frame_index:08d}"
                        f".jpg"
                    )
                )

                cv2.imwrite(

                    str(
                        preview_path
                    ),

                    preview_frame

                )

            # ====================================
            # Progress
            # ====================================

            processed_frames += 1

            if (
                processed_frames
                % 100
                == 0
            ):

                log(

                    f"{video_id} | "
                    f"processed "
                    f"{processed_frames:,} "
                    f"frames | "
                    f"faces "
                    f"{total_faces:,}"

                )

            # ====================================
            # Max frames
            # ====================================

            if (
                max_frames is not None
                and
                processed_frames
                >= max_frames
            ):
                break

            source_frame_index += 1

    finally:

        cap.release()
        csv_file.close()

    log("")
    log("=" * 70)

    log(
        f"DONE: {video_id}"
    )

    log(
        f"Processed frames: "
        f"{processed_frames:,}"
    )

    log(
        f"Detected faces: "
        f"{total_faces:,}"
    )

    log(
        f"CSV: {csv_path}"
    )

    log(
        f"Preview: "
        f"{preview_dir}"
    )

    log("=" * 70)


# ============================================================
# MAIN
# ============================================================

def main():

    parser = argparse.ArgumentParser(

        description=(
            "SCRFD Face Detection "
            "for Lip Reading Dataset"
        )

    )

    parser.add_argument(

        "--video-dir",

        default=str(
            RAW_VIDEO_DIR
        ),

        help=(
            "Input video directory"
        )

    )

    parser.add_argument(

        "--output-dir",

        default=str(
            OUTPUT_ROOT
        ),

        help=(
            "Output directory"
        )

    )

    parser.add_argument(

        "--max-frames",

        type=int,

        default=None,

        help=(
            "Maximum processed frames "
            "per video"
        )

    )

    parser.add_argument(

        "--stride",

        type=int,

        default=1,

        help=(
            "Process every N frame"
        )

    )

    parser.add_argument(

        "--preview-every",

        type=int,

        default=100,

        help=(
            "Save preview every N "
            "processed frames"
        )

    )

    parser.add_argument(

        "--confidence",

        type=float,

        default=0.5,

        help=(
            "Face confidence threshold"
        )

    )

    parser.add_argument(

        "--det-size",

        type=int,

        default=640,

        help=(
            "SCRFD detection size"
        )

    )

    parser.add_argument(

        "--cpu",

        action="store_true",

        help=(
            "Force CPU inference"
        )

    )

    args = parser.parse_args()

    video_dir = Path(
        args.video_dir
    ).resolve()

    output_dir = Path(
        args.output_dir
    ).resolve()

    output_dir.mkdir(
        parents=True,
        exist_ok=True
    )

    # ============================================
    # HEADER
    # ============================================

    log("")
    log("=" * 70)

    log(
        "SCRFD FACE DETECTION "
        "— LIP READING"
    )

    log("=" * 70)

    log(
        f"Project : "
        f"{PROJECT_ROOT}"
    )

    log(
        f"Input   : "
        f"{video_dir}"
    )

    log(
        f"Output  : "
        f"{output_dir}"
    )

    log("=" * 70)

    # ============================================
    # Detector
    # ============================================

    detector = (
        create_face_detector(

            det_size=
                args.det_size,

            force_cpu=
                args.cpu,

        )
    )

    # ============================================
    # Videos
    # ============================================

    videos = discover_videos(
        video_dir
    )

    if not videos:

        log(
            "\nไม่พบวิดีโอใน "
            f"{video_dir}"
        )

        return

    log(
        f"\nVideos found: "
        f"{len(videos)}"
    )

    # ============================================
    # Process
    # ============================================

    for video_path in videos:

        try:

            process_video(

                video_path=
                    video_path,

                detector=
                    detector,

                output_root=
                    output_dir,

                max_frames=
                    args.max_frames,

                frame_stride=
                    args.stride,

                preview_every=
                    args.preview_every,

                confidence_threshold=
                    args.confidence,

            )

        except Exception as e:

            log("")
            log("=" * 70)

            log(
                f"ERROR: "
                f"{video_path.name}"
            )

            log(
                str(e)
            )

            log("=" * 70)


# ============================================================
# ENTRY POINT
# ============================================================

if __name__ == "__main__":

    main()
```
